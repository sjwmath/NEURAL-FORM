# NEURAL FORM — 첫 RSA 실행

이 노트북은 프로젝트의 재현 가능한 실행 진입점입니다. 먼저 프로젝트 폴더에서 `run.ps1 -Install` (Windows) 또는 `bash run.sh` (Linux/WSL2)을 실행해 의존성을 설치하세요. FlyVis 공식 지원 안내는 Linux에서의 테스트를 기준으로 하므로 Windows에서는 WSL2를 권장합니다.

이 분석은 인간의 **시각 심상**을 측정하지 않습니다. 이미지를 보는 동안의 EEG와 FlyVis 모델의 시각 반응 구조를 탐색적으로 비교합니다.

```python
from pathlib import Path
import subprocess, sys
project = Path.cwd()
if not (project / 'src' / 'main.py').exists():
    project = Path('../')
assert (project / 'src' / 'main.py').exists(), 'NEURAL_FORM 프로젝트 폴더에서 노트북을 실행하세요.'
venv_python = project / ('.venv/Scripts/python.exe' if sys.platform == 'win32' else '.venv/bin/python')
subprocess.run([str(venv_python), '-m', 'src.main', '--images', '5', '--permutations', '999'], cwd=project, check=True)
```
